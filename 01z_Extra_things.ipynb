# Create a circuit with 2 qubit

```python
from qiskit import QuantumCircuit, Aer
from qiskit.visualization import plot_histogram 
from math import sqrt, pi
```

```python
## Create a quantum circuit with two qubits
qc = QuantumCircuit(2)

## Apply initialisation operation to the the qubits from a bit-string
qc.initialize('11', qc.qubits) # the order must be inverted!

## Let's view our circuit
qc.draw()  
```
