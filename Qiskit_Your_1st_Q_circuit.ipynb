## 1- Setup a quantum circuit & extract an output
* Define a quantum register  
* Mesure the qubits  
* Encode measurements in a classical register  
* Visualize the measurement results  

```python
## Import libraries
from qiskit import QuantumCircuit, assemble, Aer
from qiskit.visualization import plot_histogram
```

```python
## creates a circuit with several qubits
qc_output = QuantumCircuit(5)
#qc_output.draw()
```

```python
## Measure every qubits and store result in a classical bit
qc_output.measure_all()
```

```python
## Visualize the circuit

#qc_output.draw()
qc_output.draw(initial_state=True)
```

```python
## Run the circuit many times and plot results in a histogram (this is a trivial case)
## by default, the experiment runs 1024 times

sim = Aer.get_backend('aer_simulator') 
result = sim.run(qc_output).result()
counts = result.get_counts() 
plot_histogram(counts)
```

## 2- Encoding an input  
* A single qubit gate: the Not-gate
* binary representation

```python
qc_encode = QuantumCircuit(5) #total number of qubits n
qc_encode.x(4)  #Qubits indexing starts with 0 up to n-1
#qc_encode.draw()
```

```python
qc_encode.measure_all()
qc_encode.draw()
```

```python
sim = Aer.get_backend('aer_simulator') 
result = sim.run(qc_encode).result()
counts = result.get_counts()
plot_histogram(counts)
```

**Note:** bit string indexing in Qiskit: from left (qubit n-1, measure n, bottom wire) to right (qubit 0, measure 0, top wire)
