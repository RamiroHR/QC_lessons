# Entangling gates

## The CX gate

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(2)

# This calculates what the state vector of our qubits would be
# after passing through the circuit 'qc'
ket = Statevector(qc)

# The code below writes down the state vector.
# Since it's the last line in the cell, the cell will display it as output
ket.draw(output='latex')
```

```python
## The cx acts on two qubits, which are known as the 'control' and the 'target'. 
## Let's first try using qubit 0 as control and qubit 1 as target and see what happens.

qc.x(1)

ket = Statevector(qc)
ket.draw(output='latex')
```

```python
qc.cx(1,0)

ket = Statevector(qc)
ket.draw(output='latex')
```

```python
# Let's create a fresh quantum circuit
qc = QuantumCircuit(2)

qc.h(1)

ket = Statevector(qc)
ket.draw(output='latex')
```

```python
qc.draw()
```

```python
#The cx gate has creates entanglement!
#Syntax cx(control_qubit, target_qubit)

qc.cx(1,0)

ket = Statevector(qc)
display(ket.draw(output='latex'))
qc.draw()
```
