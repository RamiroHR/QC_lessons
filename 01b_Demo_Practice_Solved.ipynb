# Visualizing our 1-qubit Quantum Circuit
Let's now explore a more visual representation of our 1-qubit system.

## 1- Initialize state

To properly start our quantum programs we need to set the qubits in the quantum computer to start out in a defined states.  
To do so, we use the class ```QuantumCircuit``` object and the ```initialize()``` method.  
```QuantumCircuit``` objects also have a method ```draw``` that allows to see the circuit.  


````python
# import libraries to initialize a qubit in a Quantum Computer (* cloud service)
from qiskit import QuantumCircuit, Aer
from qiskit.visualization import plot_histogram 
from math import sqrt, pi
````

````python
## Create a quantum circuit with one qubit
qc = QuantumCircuit(1)

## Define initial_state as |1>, this is the column vector [0,1]
initial_state = [1,0]   

## Apply initialisation operation to the 0th qubit
qc.initialize(initial_state, 0) 

## Let's view our circuit
qc.draw()  
````

Select a specific backend hardware to execute our circuit.  
To get the results from our circuit, we use ```run()``` to execute our circuit, giving the circuit and the backend as arguments.  
We then use ```.result()``` to get the result of this:

````python
# backend selection
sim = Aer.get_backend('aer_simulator')

# save statevector at this point
qc.save_statevector()

# preset the defined circuit in the backend
result = sim.run(qc).result() 
````

from result, we can then get the final statevector using .get_statevector():

````python
# get the saved statevector
out_state = result.get_statevector()

# display statevector
print("Qubit initialized on state:")
display(out_state.draw('latex'))
````

## 2- Measure

Let’s now add a measurement to our circuit and see the result of the measurement process:

````python
qc.measure_all()

# re-draw the circuit, now with the measurement
qc.draw()
````

````python
# Execute circuit and get results
result = sim.run(qc,shots=1000).result()   ## Set to run 1000 times to have statistics

counts = result.get_counts()

plot_histogram(counts)
````

## Superposition

let’s instead put our qubit into a superposition and see what happens:  
$\frac{1}{\sqrt{2}}|0\rangle - \frac{i}{\sqrt{2}}|1\rangle$


## Initialization

````python
## Create a quantum circuit with one qubit
qc = QuantumCircuit(1)

## Define initial_state as |-i> state
initial_state = [1/sqrt(2), -1j/sqrt(2)]   

## Apply initialisation operation to the 0th qubit
qc.initialize(initial_state, 0) 

# save the statevector at this point
qc.save_statevector()

## display the circuit
qc.draw()  
````

### Verify that initialization works:

````python
# preset the defined circuit in the backend
result = sim.run(qc).result() 

# recover the saved statevector
out_state = result.get_statevector()

# display statevector
print("Qubit initialized on state:")
display(out_state.draw('latex'))
````

## Add measurement

````python
# add measurement operation.
qc.measure_all()

# re-draw the circuit, now with the measurement
qc.draw()
````

````python
# Execute circuit 1000 times and get the result
result = sim.run(qc,shots=1000).result()   ## Set to run 1000 times to have statistics

counts = result.get_counts()

# display results:
plot_histogram(counts)
````
