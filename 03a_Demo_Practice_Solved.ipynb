# Setup a quantum circuit with several qubits
* Define a multiqubit quantum register  
* Mesure all the qubits: ```measure_all()```  
* Encode measurements in a classical register and get data: ```get_counts()```
* Visualize the measurement results  

````python
## Import libraries
from qiskit import QuantumCircuit, Aer
from qiskit.visualization import plot_histogram, plot_bloch_multivector
from qiskit.quantum_info import Statevector
````

````python
## creates a circuit with 5 qubits
qc = QuantumCircuit(5)
````

````python
## Visualize all qubits in block sphere: one by one!
state = Statevector(qc)
plot_bloch_multivector(state)
````

````python
## Measure every qubits at once: stores result in a classical bit.
qc.measure_all()
````

````python
## Visualize the circuit

qc.draw(initial_state=True)
````

````python
## Run the circuit many times and plot results in a histogram (this is a trivial case)
## by default, the experiment runs 1024 times. Otherwise set argument run(.., shots=N)

# set backend
sim = Aer.get_backend('aer_simulator') 

result = sim.run(qc,shots=1000).result()

counts = result.get_counts() 

plot_histogram(counts)
````
